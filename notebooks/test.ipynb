```python
import csv
import numpy as np

r_all_results = np.zeros((8 * 10, 101, 101, 41))

for i in range(8):
    file_name = f"../sir_worker_{i:02}.npz"
    results = np.load(file_name)
    r_all_results[i * 10:(i + 1) * 10] = results['r_all']

print(r_all_results.shape)

# np.savez_compressed("r_all_results.npz", r_all_results=r_all_results)
```

```python
import matplotlib.pyplot as plt

dlamb = 0.01
lamb_values = np.arange(0, 1+dlamb, dlamb)

dgamma = 0.01
gamma_values = np.arange(0, 1+dgamma, dgamma)

gamma_idx_list = [33, 35, 37]
fig, ax = plt.subplots(1, len(gamma_idx_list), figsize=(5 * len(gamma_idx_list), 5))

for ax_idx, gamma_idx in enumerate(gamma_idx_list):
    gamma = gamma_values[gamma_idx]
    ax[ax_idx].plot(lamb_values, r_all_results[:, :, gamma_idx, -1].mean(axis=0)/10000, label=f"gamma={gamma:.2f}")
    ax[ax_idx].legend()

plt.xlabel("lambda")
plt.ylabel("R")
plt.title("R(lambda, gamma)")
plt.grid(linestyle=":")
plt.tight_layout()
plt.show()

print(r_all_results.shape)
```
